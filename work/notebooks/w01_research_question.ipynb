## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

The starter dataset contains clear signals of content exposure and decline. The code below computes three baseline numbers to confirm the signal distribution.

```python
# Load the starter dataset and compute 2-3 supporting real numbers
import pandas as pd
import os

paths_to_try = [
    "data/raw/content_refresh_anonymized.csv",
    "../../data/raw/content_refresh_anonymized.csv",
    "../data/raw/content_refresh_anonymized.csv"
]

df = None
for path in paths_to_try:
    if os.path.exists(path):
        df = pd.read_csv(path)
        break

total_pages = len(df)
declining_pages = (df['trend_direction'] == 'down').sum()
decline_pct = (declining_pages / total_pages) * 100
avg_impressions = df['impressions_90d'].mean()

print(f"1. Total Content Items in Starter Dataset: {total_pages:,}")
print(f"2. Pages Flagged with Downward Trend: {declining_pages:,} ({decline_pct:.1f}%)")
print(f"3. Average 90-Day Impressions per Page: {avg_impressions:,.2f}")
```
